# Power Plant Tolling Put Valuation Showcase

This is a public showcase notebook for an energy-derivatives valuation project. It contains only selected summaries and output figures. The full Excel-calibrated notebook and original input data are kept private.


## Project Goal

This project values a fixed-strike put option on the tolling value of a natural-gas-fired power plant. The product acts as downside protection when the plant's simulated tolling value falls below a fixed floor.


## Economic Setup

A gas-fired power plant earns value from the spark spread:

```text
Power Price - Heat Rate × Gas Price - Variable O&M
```

When this spread is positive, the plant has economic value. When power prices fall or gas prices rise, the plant's value decreases.

The put payoff is:

```text
max(K - V_toll, 0)
```

where `V_toll` is the simulated total tolling value at the put expiry date.


## Original Modeling Issue

The original course version directly simulated annual spark-spread revenue. After receiving feedback, I improved the framework to better match an energy-derivatives valuation approach.


## Improved Framework

The improved model uses a two-stage structure:

1. simulate forward power and gas prices to the put expiry date;
2. price the remaining monthly tolling values at expiry using Kirk-style spread-option logic;
3. aggregate monthly toll values;
4. compute the put payoff;
5. discount the expected payoff back to the valuation date.


## Volatility Calibration

The improved version fits Samuelson-style volatility curves for both power and gas. This captures the commodity maturity effect, where near-term contracts tend to have higher volatility than longer-dated contracts.

![Samuelson Fit](figures/samuelson_fit.png)


## Variance Mapping

The model maps total implied variance into two pieces:

1. simulation variance from valuation date to put expiry;
2. residual variance from put expiry to each delivery month.

This addresses the difference between simulating forward prices to expiry and pricing the remaining underlying toll value after expiry.


## Correlation Structure

The model uses a 36-dimensional correlated state vector: 12 peak power forwards, 12 off-peak power forwards, and 12 gas forwards. The explicit correlation structure is important because the plant's economics depend on the spark spread, not power or gas alone.

![Correlation Heatmap](figures/correlation_heatmap.png)


## Valuation Results

The base underlying toll value at put expiry was approximately **$80.0 million**.

The fixed strike was set to 80% of the base toll value, or approximately **$64.0 million**.

The estimated present value of the put was approximately **$4.63 million**, with an exercise probability of about **34.9%**.

The Monte Carlo standard error was approximately **$0.118 million**, giving a 95% confidence interval of approximately **$4.40 million to $4.86 million**.

Power-gas correlation has a material impact on valuation. The improved notebook tests several correlation assumptions and shows that the put value changes meaningfully as correlation changes.

## Sensitivity and Stress Tests

The put is most sensitive to downside power-price risk. Lower power prices, higher gas prices, worse heat rate, and higher forced outages increase the value of protection.

![Sensitivity Tests](figures/sensitivity_tests.png)


## Monthly Risk Measures

The improved version reports monthly Deltas and Vegas.

Power Deltas are negative because higher power prices increase the plant's tolling value and reduce the value of the put.

Gas Deltas are positive because higher gas prices reduce spark spreads and increase the value of downside protection.

## Important Limitations

This is an educational valuation model, not a production trading model.

Several assumptions remain simplified:

- off-peak power forwards are approximated from peak forwards;
- off-peak volatility is approximated from peak volatility;
- dispatch is simplified and does not fully model unit commitment;
- the correlation structure is partly assumption-based;
- Samuelson calibration can produce high short-maturity parameters, so the fitted curves should be interpreted as a variance-mapping framework rather than exact market quotes.
